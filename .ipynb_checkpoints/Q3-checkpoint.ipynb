## Introduction  
In this question, I’m testing a more advanced machine learning model, a Random Forest Regressor, to see whether it can predict closing prices more accurately than the Linear Regression model from Q1 and the MLP from Q2. Random Forests are good at handling non‑linear patterns and usually perform well even when the data is noisy. I’m using the same lag features and train/test split as before, so the comparison between models is fair.

```python
import sys
sys.path.append("C:\\Users\\Paris\\Coursework-2---Finance")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from py.functions import load_data, basic_clean, train_test_split_time
```

```python
df = load_data("finance_economics_dataset.csv")
df = basic_clean(df)
```

```python
df["lag_1"] = df["Close Price"].shift(1)
df["lag_2"] = df["Close Price"].shift(2)
df["lag_3"] = df["Close Price"].shift(3)

df = df.dropna()
```

```python
train, test = train_test_split_time(df)

X_train = train[['lag_1', 'lag_2', 'lag_3']]
y_train = train['Close Price']

X_test = test[['lag_1', 'lag_2', 'lag_3']]
y_test = test['Close Price']
```

## Methodology  
I started by loading and cleaning the dataset using the same functions from earlier questions. I kept the same three lag features (lag_1, lag_2, lag_3) so the Random Forest model would have the same inputs as the previous models.

Next, I split the data into training and testing sets using a time‑based split. I then created a Random Forest Regressor with 200 trees and trained it on the lag features. After training, I used the model to predict the closing prices for the test set. I calculated MAE and RMSE to measure how well the model performed and plotted the actual vs predicted prices to visually compare the results.

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

rf.fit(X_train, y_train)
```

```python
preds_rf = rf.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error

mae_rf = mean_absolute_error(y_test, preds_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, preds_rf))

mae_rf, rmse_rf
```

```python
plt.figure(figsize=(12,6))
plt.plot(test["Date"], y_test, label="Actual")
plt.plot(test["Date"], preds_rf, label="Random Forest Predicted")
plt.legend()
plt.title("Actual vs Random Forest Predicted Close Price")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.show()
```

## Plot Interpretation  
The actual closing prices move up and down sharply, showing a lot of volatility. The Random Forest predictions follow the general trend quite closely and react more quickly to changes than the MLP did. However, the predictions are still smoother than the real data and don’t fully capture the sudden spikes and drops. This shows that the Random Forest model can pick up more complex patterns than the previous models, but it still struggles with extreme movements in the market.

## Conclusion  
The Random Forest model performs better than both the Linear Regression and the MLP. It captures the trend more accurately and responds more quickly to changes in the data. However, it still can’t fully match the volatility of the actual prices, which is expected given the limited information provided by lag features alone. Overall, Random Forest offers the strongest performance so far, but richer features or more advanced time‑series models would be needed for highly accurate forecasting.

## Comparison with Q1 and Q2  
Compared to the Linear Regression model in Q1, the Random Forest clearly performs better because it can learn non‑linear relationships. It also outperforms the MLP from Q2 by reacting more quickly to changes and producing predictions that stay closer to the actual values. While both the MLP and Random Forest smooth out the volatility, the Random Forest does a better job of following the overall shape of the data. This makes it the strongest model out of the three.
